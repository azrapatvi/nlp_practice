```python
import nltk
nltk.download('punkt_tab')
```

```python

from nltk.tokenize import word_tokenize



text='She is a beautiful girl and she studies very hard.'
tokens=word_tokenize(text)

print(tokens)
```

```python
from nltk.stem import SnowballStemmer
from nltk.corpus import stopwords

nltk.download('stopwords')
stop_words=set(stopwords.words('english'))
```

```python
stemmer=nltk.LancasterStemmer()
```

```python


text='She is a beautiful girl and she studies very hard.'

text=text.lower()
print(text)

tokens=word_tokenize(text)
print(tokens)

text=[stemmer.stem(word) for word in tokens if word not in stop_words]
print(text)
```

```python
nltk.download('averaged_perceptron_tagger_eng')
```

```python
from nltk import pos_tag

text_with_pos=pos_tag(text)

print(text_with_pos)
```

```python
text="""
Artificial Intelligence is rapidly changing the way people learn, work, communicate, and solve problems in today's world. Students are learning new technologies every day, and they are becoming more interested in machine learning, natural language processing, and data science. A computer can analyze thousands of documents within a few seconds, identify important information, and provide useful results. However, computers do not naturally understand human language in the same way that humans do. Therefore, researchers have developed various techniques to process text efficiently. Text preprocessing is one of the most important steps in Natural Language Processing because raw text often contains unnecessary words, punctuation, numbers, symbols, and different forms of the same word. For example, words such as playing, played, plays, and player may need to be converted into a common form depending on the preprocessing technique being used. Similarly, words like studies, studying, studied, and study can be handled using stemming or lemmatization. Stemming usually removes prefixes or suffixes to produce a root-like form, while lemmatization tries to return the meaningful dictionary form of a word. Stopword removal can also eliminate frequently occurring words such as the, is, are, was, were, and, or, but, in, on, at, and to when they do not provide useful information for a particular task. After preprocessing, the cleaned text can be transformed into numerical features and given to machine learning models. These models can then learn patterns from the text and make predictions, classifications, or other useful decisions. In recent years, researchers have been developing powerful language models that can understand, generate, summarize, translate, and analyze human language. As technology continues to improve, natural language processing will become even more useful in education, healthcare, business, customer service, and many other areas.
"""

print(text)
```

```python
from nltk.tokenize import sent_tokenize

text=text.lower()
sentences=sent_tokenize(text)

print(sentences)
```

```python
stemmer=nltk.PorterStemmer()
```

```python
for i in range(len(sentences)):
  words=word_tokenize(sentences[i])

  words=[stemmer.stem(word) for word in words if word not in stop_words]

  print(pos_tag(words))
```
