```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.shape
```

```python
df.columns=df.columns.str.lower()
```

```python
X=df['text']
y=df['language']
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=42,stratify=y)
```

```python
X_train
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf=TfidfVectorizer()
```

```python
X_train=tfidf.fit_transform(X_train)
X_test=tfidf.transform(X_test)
```

```python
from sklearn.naive_bayes import MultinomialNB

model_without_weights=MultinomialNB()
```

```python
model_without_weights
```

```python
model_without_weights.fit(X_train,y_train)
```

```python
y_pred=model_without_weights.predict(X_test)

y_pred
```

```python
from sklearn.metrics import accuracy_score

print(accuracy_score(y_test,y_pred))
```

```python
from sklearn.naive_bayes import MultinomialNB

model=MultinomialNB()
```

```python
from sklearn.utils.class_weight import compute_sample_weight

sample_weights=compute_sample_weight(
    class_weight='balanced',
    y=y_train
)
```

```python
model.fit(X_train, y_train, sample_weight=sample_weights)
```

```python
y_pred=model.predict(X_test)

y_pred
```

```python
from sklearn.metrics import accuracy_score

print(accuracy_score(y_test,y_pred))
```

```python
new_text = ["आज मौसम बहुत अच्छा है और मैं अपने दोस्तों के साथ बाहर जा रहा हूँ।"]

new_text_tfidf = tfidf.transform(new_text)

prediction = model.predict(new_text_tfidf)

print(prediction)
```
