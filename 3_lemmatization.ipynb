```python
import nltk

nltk.download('wordnet')
```

```python

# the output we get after stemming is called stem and output we get from lemmatiation is lemma
# the diffenrece between stemmign and lemmatization is:
# stemmign will reduce the words to its base form by simply removing the prefixes and suffixes
# while the lemmatization is not like thatt it used dictionry of wordss

# and after lemmatization the word we will get is alwayss a valid word


# postag
# n->Noun
# v->verb
# a->adjective
# r->adverb

from nltk.stem import WordNetLemmatizer

lemmatizer=WordNetLemmatizer()

lemmatizer.lemmatize("going")
```

```python
# verb when there is an action or state like running going walking ran  studied
lemmatizer.lemmatize("going",pos='v')
```

```python
# noun - person place or thing she he them

lemmatizer.lemmatize("she",pos='n')
```

```python
# adjective - for words that descirbes the noun like beautiful girl ,cute cat,big house,better

lemmatizer.lemmatize("better",pos='a')
```

```python
#adverb- whne the wrods describes how an action happens

lemmatizer.lemmatize('quickly',pos='r')
```
