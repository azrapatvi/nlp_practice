```python
import nltk # natural langaugae tool kit a python librar for nlp

nltk.download('punkt_tab') # we need to install thiss beause nltk wont understandd without this that from where to split the the data

# punkt_tab contains data used for nltk tokenizer
```

##### sentence tokenization

```python
corpus="I love Python. Python is easy to learn. NLP is interesting."
```

```python
corpus
```

```python
from nltk.tokenize import sent_tokenize

tokens=sent_tokenize(corpus)
print(tokens)
```

```python
corpus
```

##### word tokenization

```python
from nltk.tokenize import word_tokenize

word_tokens=word_tokenize(corpus)

print(word_tokens)
```

```python
from nltk.tokenize import wordpunct_tokenize

sent1="Azra's tutorials."
wordpunct_tokenize(sent1) # so this will seprate out the puchtuation marks s well
```

```python
from nltk.tokenize import TreebankWordTokenizer

sent1="Azra's tutorials. how are you, hope you are doing well."
tokenizer=TreebankWordTokenizer() # so this will not seprate out the the full stop which is in between the sentences but will create a separate token for fullstop when the full stop is in last sentence last word

tokenizer.tokenize(sent1)
```
